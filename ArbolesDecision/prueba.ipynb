```python
def fibo (next) :
    a, b = 0, 1
    for _ in range(next):
        yield a
        a, b = b, a + b

fibo_gen = fibo(2)
for num in fibo_gen:
    print(num)
```

```python
def grafica_seno():
    import numpy as np
    import matplotlib.pyplot as plt

    x = np.linspace(0, 2 * np.pi, 100)
    y = np.cos(x)
    y1 = np.sin(x)
    y2 = np.tan(x)
    y3 = np.sinh(x)
    y4 = np.cosh(x)
    y5 = np.tanh(x)
    plt.plot(x, y, label='coseno', color='orange')
    plt.plot(x, y1, label='seno', color='blue')
    plt.plot(x, y2, label='tangente', color='green')
    plt.plot(x, y3, label='seno hiperbólico', color='purple')
    plt.plot(x, y4, label='coseno hiperbólico', color='red')
    plt.plot(x, y5, label='tangente hiperbólica', color='brown')
    plt.title('Gráfica de las funciones seno / coseno / tangente / seno hiperbólico / coseno hiperbólico / tangente hiperbólica')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid()
    plt.legend()
    plt.show()
    
grafica_seno()
```

```python
def grafico_3d():
    import numpy as np
    import matplotlib.pyplot as plt
    from mpl_toolkits.mplot3d import Axes3D  # necesario para 3D

    # Crear dominio
    x = np.linspace(-5, 5, 200)
    y = np.linspace(-5, 5, 200)
    X, Y = np.meshgrid(x, y)

    # Función
    Z = np.sin(np.sqrt(X**2 + Y**2))

    # Figura
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    # Superficie
    superficie = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

    # Etiquetas
    ax.set_title('Gráfico 3D de z = sin(√(x² + y²))')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

    # Barra de color
    fig.colorbar(superficie, ax=ax, shrink=0.5, aspect=10)

    plt.show()
grafico_3d()
```
